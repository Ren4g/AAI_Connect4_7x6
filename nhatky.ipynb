Phương pháp (viết một lần, sửa khi đổi): luật chơi, từng agent, cách chạy đánh giá (khai cuộc ngẫu nhiên, đổi ghế, ngân sách thời gian, khoảng tin cậy), cách dùng solver. Phần này gần như ổn định và chiếm nhiều công nhất khi viết sau.

connect4 7x6 - 7 cột 6 hàng 42 ô, 2 người đối kháng, thả quân từ trên cao xuống bàn cờ đứng, ai nối được 4 quân trước là thắng. 

Nhật ký kết quả (thêm vào cuối, theo ngày): mỗi lần chạy ghi lại ngày, lệnh/seed, file kết quả, con số chính, và điều bạn rút ra hoặc còn nghi ngờ. Đừng sửa lại các dòng cũ, nhất là khi kết quả sau mâu thuẫn trước. Việc thừa nhận và giải thích sự thay đổi cũng là nội dung báo cáo.

Việc còn mở: những giả thuyết chưa kiểm chứng (ví dụ nguyên nhân CNN yếu) và việc cần làm.
